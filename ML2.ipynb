```python
//Modeling Processing in machine learning//
//Boston Houseing Price Prediction//
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/Boston House Price Data.csv")
data
```

```python
plt.scatter(data["RM"], data["PRICE"])
plt.xlabel("Average number of rooms(RM)")
plt.ylabel("House PRICE")
plt.title("RM VS PRICE")
plt.show()
```

```python
correlation=data.corr(numeric_only=True)
sns.heatmap(correlation, annot=True)
plt.title("correlation Heatmap")
plt.show()
```

```python
x=data.drop("PRICE" , axis=1)
y=data["PRICE"]
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=42)

print("Training data:", x_train.shape)
print("Training data:", x_test.shape)
print("Training data:", y_train.shape)
print("Training data:", y_test.shape)

```

```python
model=LinearRegression()
model.fit(x_train, y_train)

print("Model Trained Successfully..!!")
```

```python
model.fit(x_train, y_train)
```

```python
model.predict(x_test)
```

```python
y_pred=model.predict(x_test)
```

```python
print("Prediction Value:" ,y_pred[0:4])
```

```python
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```
