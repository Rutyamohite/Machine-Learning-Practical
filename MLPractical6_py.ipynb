```python
//KNN Diabetes Prediction//

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

data = pd.read_csv("diabetes.csv")
data


```

```python
import matplotlib.pyplot as plt

# Histogram - Glucose Distribution
plt.figure(figsize=(6, 4))

plt.hist(data["Glucose"], bins=20)

plt.xlabel("Glucose")
plt.ylabel("Frequency")
plt.title("Glucose Distribution")

plt.show()
```

```python
#Scatter Plot - Glucose vs BMI

plt.figure(figsize=(6, 4))

plt.scatter(data["Glucose"], data["BMI"])

plt.xlabel("Glucose")
plt.ylabel("BMI")
plt.title("Glucose vs BMI")

plt.show()
```

```python
# Bar Chart - Diabetes Outcome
plt.figure(figsize=(5, 4))

sns.countplot(data=data, x="Outcome")

plt.xlabel("Outcome")
plt.ylabel("Number of Patients")
plt.title("Diabetes Outcome")

plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(data.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
x = data.drop("Outcome", axis=1)
y = data["Outcome"]
```

```python
X_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(x_test)
```

```python
for actual, predicted in zip(y_test, y_pred):
  if actual == predicted:
    print("Correct Prediction: ", "Actual=", actual, "Predicted =", predicted)
  else:
    print("Wrong Prediction: ", "Actual=", actual, "Predicted =", predicted)
```
